```python
from hw2fintools import gurufocus as gf
import hw2fintools as ft
import os
from dotenv import load_dotenv

load_dotenv()
# Common Variables & Dataframes
path_stockdata = os.path.join(os.environ.get('judgement_day'), 'Data--StockWatchList')
df_0 = ft.get_df_stock_watch_list()
ticker_list = df_0['Ticker']
gt = os.environ.get('guru_token')

```

```python
for ticker in ticker_list:
    path_ticker = os.path.join(path_stockdata, ticker.upper())

    print(f'Start--{ticker}')

    an_list = []

    for item in os.listdir(path_ticker):
        if 'gf-raw-financials_annual_history' in item and item.endswith('.csv') and not item.startswith('._'):
            an_list.append(item)

    an_list_sorted = sorted(an_list, reverse=True)
    an_current = an_list_sorted[0]

    an_path = os.path.join(path_ticker, an_current)

    param_list = []

    for item in os.listdir(path_ticker):
        if 'gf-raw-financials_parameters' in item and item.endswith('.csv') and not item.startswith('._'):
            param_list.append(item)

    param_list_sorted = sorted(param_list, reverse=True)
    param_current = param_list_sorted[0]

    param_path = os.path.join(path_ticker, param_current)

    out_df = gf.annual_10K_hist_s1v1(fin_file_path=an_path, param_file_path=param_path)

    out_df.to_csv(os.path.join(path_ticker, ticker.upper() + '--Annual_10K-s1v1.csv'))

    print(f'Stop--{ticker}')
```
