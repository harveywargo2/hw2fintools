## Start

```python
import os
from dotenv import load_dotenv
import pandas as pd
import datetime
import plotly.graph_objects as go


load_dotenv()
ticker = ''
path_stockdata = os.path.join(os.path.join(os.environ.get('judgement_day'), 'Data--StockWatchList'), ticker)
path_analysis_csv = os.path.join(path_stockdata, f'{ticker}--Analysis_RevGro-s1v1.csv')

today = datetime.date.today()
cutoff = today.year - 20
cutoff10 = today.year - 10
cutoff5 = today.year - 5
cutoff3 = today.year - 3
```

## Data Imports & Cleaning

```python
%%capture

df0 = pd.read_csv(os.path.join(path_stockdata, f'{ticker}--Annual_10K-s1v1.csv'), index_col=0)
df0 = df0.dropna(subset=['FiscalYear'])
df0[['FiscalYear', 'FiscalMonth']] = df0[['FiscalYear', 'FiscalMonth']].astype(int)
```

```python
df0
```

```python
%%capture

growth_df0 = df0[['FiscalYear', 'Revenue']]
growth_df0['RevGro'] = round(growth_df0['Revenue'].pct_change() , 2)
```

```python
growth_df0
```

## Revenue Growth

```python
revgro_fig = go.Figure(data=[
   go.Bar(name='BasicChg', x=growth_df0['FiscalYear'].tail(20), y=growth_df0['RevGro'].tail(20) * 100, offsetgroup=1, marker_color='blue'),
])
revgro_fig.update_xaxes(dtick=1)
revgro_fig.update_layout(yaxis_title='% Change', xaxis_title='FiscalYear', title='Year20 RevGro', template='plotly_dark')
revgro_fig.show()
```

```python
revgro_fig1 = go.Figure(data=[
   go.Bar(name='Rev', x=growth_df0['FiscalYear'].tail(20), y=growth_df0['Revenue'].tail(20), offsetgroup=1, marker_color='green'),
])
revgro_fig1.update_xaxes(dtick=1)
revgro_fig1.update_layout(yaxis_title='Dollars Mil', xaxis_title='FiscalYear', title='20 Year Revenue', template='plotly_dark')
revgro_fig1.show()
```

```python
revgro_mean20 = growth_df0['RevGro'].tail(20).mean()
revgro_median20 = growth_df0['RevGro'].tail(20).median()

revgro_mean10 = growth_df0['RevGro'].tail(10).mean()
revgro_median10 = growth_df0['RevGro'].tail(10).median()

revgro_mean5 = growth_df0['RevGro'].tail(5).mean()
revgro_median5 = growth_df0['RevGro'].tail(5).median()

revgro_mean3 = growth_df0['RevGro'].tail(3).mean()
revgro_median3 = growth_df0['RevGro'].tail(3).median()

revgro_lst = growth_df0['RevGro'].iloc[-1]

print(f'Mean20Year RevGro: {round(revgro_mean20 * 100, 2)}%')
print(f'Median20Year RevGro: {round(revgro_median20 * 100, 2)}%')
print()
print(f'Mean10Year RevGro: {round(revgro_mean10 * 100, 2)}%')
print(f'Median10Year RevGro: {round(revgro_median10 * 100, 2)}%')
print()
print(f'Mean5Year RevGro: {round(revgro_mean5 * 100, 2)}%')
print(f'Median5Year RevGro: {round(revgro_median5 * 100, 2)}%')
print()
print(f'Mean3Year RevGro: {round(revgro_mean3 * 100, 2)}%')
print(f'Median3Year RevGro: {round(revgro_median3 * 100, 2)}%')
print()
print(f'Last RevGro: {round(revgro_lst * 100, 2)}%')
```

## Analysis Output

```python
metrics_json = {
    "analysis_type": "growth_revenue",
    "analysis_date": today.strftime('%Y-%m-%d'),
    "last_df_date": df0['FiscalYear'].iloc[-1],
    "mean20yr_revgro": revgro_mean20,
    "median20yr_revgro": revgro_median20,
    "mean10yr_revgro": revgro_mean10,
    "median10yr_revgro": revgro_median10,
    "mean5yr_revgro": revgro_mean5,
    "median5yr_revgro": revgro_median5,
    "mean3yr_revgro": revgro_mean3,
    "median3yr_revgro": revgro_median3,
    "last_revgro": revgro_lst,
}

metrics_json
```

```python
metrics_df = pd.DataFrame([metrics_json])

if os.path.isfile(path_analysis_csv):
    metrics_df.to_csv(path_analysis_csv, mode='a', header=False, index=False)
else:
    metrics_df.to_csv(path_analysis_csv, mode='w', header=True, index=False)

metrics_df
```
