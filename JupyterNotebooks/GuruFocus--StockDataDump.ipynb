```python
import os

import pandas as pd
from dotenv import load_dotenv
from pathlib import Path
import hw2fintools
import hw2fintools.gurufocus as gf
from datetime import datetime


load_dotenv()

# Common Variables & Dataframes
path_jd = os.environ.get('judgement_day')
path_stockdata = os.path.join(path_jd, 'StockData')
df_0 = hw2fintools.get_df_stock_watch_list()
ticker_list = df_0['Ticker']
gt = os.environ.get('guru_token')
today = datetime.today().strftime('%Y-%m-%d')

```

```python
# Directory Check & Update
for item in ticker_list:
    dir_path = Path(os.path.join(path_stockdata, item))
    os.makedirs(dir_path, exist_ok=True)

    fin = gf.FinancialHistory(token=gt, ticker=item)
    div = gf.DividendHistory(token=gt, ticker=item)
    price = gf.PriceHistory(token=gt, ticker=item)

    fin_parameters = fin.data_template_parameters
    fin_paramerters_df = pd.DataFrame.from_dict(fin_parameters, orient='index')
    fin_paramerters_df.to_csv(os.path.join(dir_path, today + '--' + item.upper() +  '--gf-financials_parameters.csv'))

    annual_df = fin.raw_df_annuals
    annual_df.to_csv(os.path.join(dir_path, today + '--' + item.upper() + '--gf-financials_annual_history' + '.csv'))

    quarterly_df = fin.raw_df_quarterly
    quarterly_df.to_csv(os.path.join(dir_path, today + '--' + item.upper() + '--gf-financials_quarterly_history' + '.csv'))

    div_df = div.raw_df
    div_df.to_csv(os.path.join(dir_path, today + '--' + item.upper() + '--gf-dividend_history' + '.csv'))

    price_df = price.raw_df
    price_df.to_csv(os.path.join(dir_path, today + '--' + item.upper() + '--gf-price_history' + '.csv'))
```
