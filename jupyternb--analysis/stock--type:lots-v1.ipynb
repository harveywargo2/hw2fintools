```python
import os
from dotenv import load_dotenv
import pandas as pd
import datetime


load_dotenv()
ticker = ''
path_stockdata = os.path.join(os.path.join(os.environ.get('judgement_day'), 'Data--StockWatchList'), ticker)
path_analysis_csv = os.path.join(path_stockdata, f'{ticker}--Analysis_LotResults-s1v1.csv')

today = datetime.date.today()
cutoff = today.year - 20
cutoff10 = today.year - 10
cutoff5 = today.year - 5
cutoff3 = today.year - 3

current_price = 213
```

```python


df0 = pd.read_csv(os.path.join(path_stockdata, f'{ticker}--LotDetails.csv'))
df0['Date'] = pd.to_datetime(df0['Date'])

df0
```

```python
df1 = pd.read_csv(os.path.join(path_stockdata, f'{ticker}--Div_History-s1v1.csv'), index_col=0)
df2 = pd.DataFrame()
df2['ExDate'] = pd.to_datetime(df1['ex_date'])
df2['DivAmount'] = df1['amount']
df2
```

```python
merged_df = df0.merge(df2, how='cross')
merged_df
```

```python
merged_df = merged_df[merged_df['ExDate'] > merged_df['Date']]
merged_df
```

```python
lot_df = merged_df.groupby(['Date', 'Cost'])['DivAmount'].sum().reset_index()
lot_df
```

```python
lot_df['CurrentPrice'] = current_price
lot_df['TotalReturn'] = lot_df['DivAmount'] + lot_df['CurrentPrice']
lot_df['Periods'] = today.year - lot_df['Date'].dt.year
lot_df['CAGR'] = round((((lot_df['TotalReturn'] / lot_df['Cost']) ** (1 / lot_df['Periods'])) - 1) * 100, 2)
lot_df
```
