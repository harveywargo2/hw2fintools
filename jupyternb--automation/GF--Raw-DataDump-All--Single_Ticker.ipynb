```python
import os
import json
import pandas as pd
from dotenv import load_dotenv
from pathlib import Path
import hw2fintools
import hw2fintools.gurufocus as gf
from datetime import datetime


load_dotenv()

# Common Variables & Dataframes
path_jd = os.environ.get('judgement_day')
path_raw_data = os.path.join(path_jd, 'StockData')
df_0 = hw2fintools.get_df_stock_watch_list()
gt = os.environ.get('guru_token')
today = datetime.today().strftime('%Y-%m-%d')
```

```python
ticker = input('Enter ticker symbol: ')
print('Ticker is: ', ticker)
```

```python
dir_path = Path(os.path.join(path_raw_data, ticker.upper()))
os.makedirs(dir_path, exist_ok=True)

fin = gf.FinancialHistory(token=gt, ticker=ticker)
div = gf.DividendHistory(token=gt, ticker=ticker)
price = gf.PriceHistory(token=gt, ticker=ticker)
stock_summary = gf.StockSummary(token=gt, ticker=ticker)
segments = gf.SegmentData(token=gt, ticker=ticker)
operations = gf.OperatingData(token=gt, ticker=ticker)

fin_parameters = fin.data_template_parameters
fin_parameters_df = pd.DataFrame.from_dict(fin_parameters, orient='index')
fin_parameters_df.to_csv(os.path.join(dir_path, today + '--' + ticker.upper() +  '--gf-raw-financials_parameters.csv'))

annual_df = fin.raw_df_annuals
annual_df.to_csv(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-financials_annual_history.csv'))

quarterly_df = fin.raw_df_quarterly
quarterly_df.to_csv(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-financials_quarterly_history.csv'))

div_df = div.raw_df
div_df.to_csv(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-dividend_history.csv'))

price_df = price.raw_df
price_df.to_csv(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-price_history.csv'))

stock_summary_dict = stock_summary.raw_data
with open(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-stock_summary.json'), 'w') as json_file:
    json.dump(stock_summary_dict, json_file)

segment_dict = segments.raw_data
with open(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-segments_data.json'), 'w') as json_file:
    json.dump(stock_summary_dict, json_file)

operation_dict = operations.raw_data
with open(os.path.join(dir_path, today + '--' + ticker.upper() + '--gf-raw-operations_data.json'), 'w') as json_file:
    json.dump(stock_summary_dict, json_file)
```
