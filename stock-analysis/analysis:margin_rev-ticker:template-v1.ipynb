## Start

```python
import os
from dotenv import load_dotenv
import pandas as pd
import plotly.graph_objects as go
import datetime

load_dotenv()
ticker = ''
path_stockdata = os.path.join(os.path.join(os.environ.get('judgement_day'), 'Data--StockWatchList'), ticker)
path_analysis_csv = os.path.join(path_stockdata, f'{ticker}--Analysis_RevMargins-s1v1.csv')

today = datetime.date.today()
cutoff = today.year - 20
cutoff10 = today.year - 10
cutoff5 = today.year - 5
cutoff3 = today.year - 3
```

## Data Imports & Cleaning

```python
%%capture

df0 = pd.read_csv(os.path.join(path_stockdata, f'{ticker}--Annual_10K-s1v1.csv'), index_col=0)
df0 = df0.dropna(subset=['FiscalYear'])
df0[['FiscalYear', 'FiscalMonth']] = df0[['FiscalYear', 'FiscalMonth']].astype(int)
```

```python
df0
```

## Revenue Margins


```python
%%capture

margin_df = df0[['FiscalYear', 'Revenue', 'GrossProfit', 'OperatingIncome', 'NetIncome', 'OpCash', 'FreeCash', 'CAPEX']]
margin_df['GPM'] = margin_df['GrossProfit'] / margin_df['Revenue']
margin_df['OPM'] = margin_df['OperatingIncome'] / margin_df['Revenue']
margin_df['NPM'] = margin_df['NetIncome'] / margin_df['Revenue']
margin_df['OCM'] = margin_df['OpCash'] / margin_df['Revenue']
margin_df['FCM'] = margin_df['FreeCash'] / margin_df['Revenue']
margin_df['CAPEXM'] = margin_df['CAPEX'] / margin_df['Revenue']
```

```python
margin_df.tail(20)
```

## Profit Margins

```python
profit_fig1 = go.Figure(data=[
    go.Bar(name='GPM', x=margin_df['FiscalYear'].tail(20), y=round(margin_df['GPM'].tail(20) * 100, 2), offsetgroup=1, marker_color='blue'),
    go.Bar(name='OPM', x=margin_df['FiscalYear'].tail(20), y=round(margin_df['OPM'].tail(20) * 100, 2), offsetgroup=2, marker_color='yellow'),
    go.Bar(name='NPM', x=margin_df['FiscalYear'].tail(20), y=round(margin_df['NPM'].tail(20) * 100, 2), offsetgroup=3, marker_color='green'),

])
profit_fig1.update_xaxes(dtick=1)
profit_fig1.update_layout(yaxis_title='Margin %', xaxis_title='FiscalYear', title='Profit Margins', template='plotly_dark')
profit_fig1.show()
```

```python
gpm_mean20 = margin_df['GPM'].tail(20).mean()
gpm_median20 = margin_df['GPM'].tail(20).median()
gpm_mean10 = margin_df['GPM'].tail(10).mean()
gpm_median10 = margin_df['GPM'].tail(10).median()
gpm_mean5 = margin_df['GPM'].tail(5).mean()
gpm_median5 = margin_df['GPM'].tail(5).median()
gpm_mean3 = margin_df['GPM'].tail(3).mean()
gpm_median3 = margin_df['GPM'].tail(3).median()

print(f'Mean20Year GPM: {round(gpm_mean20 * 100, 2)}%')
print(f'Median20Year GPM: {round(gpm_median20 * 100, 2)}%')
print()
print(f'Mean10Year GPM: {round(gpm_mean10 * 100, 2)}%')
print(f'Median10Year GPM: {round(gpm_median10 * 100, 2)}%')
print()
print(f'Mean5Year GPM: {round(gpm_mean5 * 100, 2)}%')
print(f'Median5Year GPM: {round(gpm_median5 * 100, 2)}%')
print()
print(f'Mean3Year GPM: {round(gpm_mean3 * 100, 2)}%')
print(f'Median3Year GPM: {round(gpm_median3 * 100, 2)}%')
```

```python
opm_mean20 = margin_df['OPM'].tail(20).mean()
opm_median20 = margin_df['OPM'].tail(20).median()
opm_mean10 = margin_df['OPM'].tail(10).mean()
opm_median10 = margin_df['OPM'].tail(10).median()
opm_mean5 = margin_df['OPM'].tail(5).mean()
opm_median5 = margin_df['OPM'].tail(5).median()
opm_mean3 = margin_df['OPM'].tail(3).mean()
opm_median3 = margin_df['OPM'].tail(3).median()

print(f'Mean20Year OPM: {round(opm_mean20 * 100, 2)}%')
print(f'Median20Year OPM: {round(opm_median20 * 100, 2)}%')
print()
print(f'Mean10Year OPM: {round(opm_mean10 * 100, 2)}%')
print(f'Median10Year GPM: {round(opm_median10 * 100, 2)}%')
print()
print(f'Mean5Year OPM: {round(opm_mean5 * 100, 2)}%')
print(f'Median5Year OPM: {round(opm_median5 * 100, 2)}%')
print()
print(f'Mean3Year OPM: {round(opm_mean3 * 100, 2)}%')
print(f'Median3Year OPM: {round(opm_median3 * 100, 2)}%')
```

```python
npm_mean20 = margin_df['NPM'].tail(20).mean()
npm_median20 = margin_df['NPM'].tail(20).median()
npm_mean10 = margin_df['NPM'].tail(10).mean()
npm_median10 = margin_df['NPM'].tail(10).median()
npm_mean5 = margin_df['NPM'].tail(5).mean()
npm_median5 = margin_df['NPM'].tail(5).median()
npm_mean3 = margin_df['NPM'].tail(3).mean()
npm_median3 = margin_df['NPM'].tail(3).median()

print(f'Mean20Year NPM: {round(npm_mean20 * 100, 2)}%')
print(f'Median20Year NPM: {round(npm_median20 * 100, 2)}%')
print()
print(f'Mean10Year NPM: {round(npm_mean10 * 100, 2)}%')
print(f'Median10Year NPM: {round(npm_median10 * 100, 2)}%')
print()
print(f'Mean5Year NPM: {round(npm_mean5 * 100, 2)}%')
print(f'Median5Year NPM: {round(npm_median5 * 100, 2)}%')
print()
print(f'Mean3Year NPM: {round(npm_mean3 * 100, 2)}%')
print(f'Median3Year NPM: {round(npm_median3 * 100, 2)}%')
```
